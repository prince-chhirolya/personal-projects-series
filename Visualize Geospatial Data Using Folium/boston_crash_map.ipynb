#### Task 2.5: Add Mouse Position

```python
boston_coordinate = [42.361145, -71.057083]
boston_map = folium.Map(location=boston_coordinate, zoom_start=10)
formatter = "function(num) {return L.Util.formatNum(num, 5);};"
mouse_position = MousePosition(
    position='topright',
    separator=' Long: ',
    empty_string='NaN',
    lng_first=False,
    num_digits=20,
    prefix='Lat:',
    lat_formatter=formatter,
    lng_formatter=formatter,
).add_to(boston_map)
boston_map
```

#### Task 3.1: Create Heat Map

# Task 4: Create Circle Markers of all Accident Locations

```python
boston_map = folium.Map(location=boston_coordinate,
        zoom_start=12,
        zoom_control=True
        )
colors = {"ped": "green", "bike": "blue", "mv": "red"}
sample = df.head(1000)
for index, rows in sample.iterrows():
    mode = rows.mode_type
    color = colors[mode]
    folium.CircleMarker(
        location = [rows.lat, rows.long],
        radius = 5,
        fill = True,
        fill_color = color,
        fill_opacity = 0.5).add_to(boston_map)
boston_map
```

# Task 5: Customized Markers for different Accidents

```python
boston_map = folium.Map(location=boston_coordinate,
            zoom_start=13,
            zoom_control=True,
            scrollWheelZoom=False,
            dragging=True)
values = {
    "ped": {"color": "red", "icon": "person"},
    "bike": {"color": "green", "icon": "bicycle"},
    "mv": {"color": "blue", "icon": "car"}
}
sample = df.head(500)
for index, rows in sample.iterrows():
    mode = rows.mode_type
    folium.Marker(
        location = [rows.lat, rows.long],
        icon=folium.Icon(color=values[mode]['color'], prefix='fa',icon=values[mode]['icon'])).add_to(boston_map)

boston_map
```

# Task 10: Create Route on Map

```python
ox.settings.log_console=True
ox.settings.use_cache=True
start = (42.3546, -71.0708)               
end = (42.3626, -71.0688)

place = 'Boston, Massachusetts, United States'
mode = 'drive'        
optimizer = 'time'      
graph = ox.graph_from_place(place, network_type = mode)
orig_node = ox.distance.nearest_nodes(graph, start[1], start[0])
dest_node = ox.distance.nearest_nodes(graph, end[1], end[0])
shortest_route = nx.shortest_path(graph, orig_node, dest_node, weight=optimizer)

route_map = ox.plot_route_folium(graph, shortest_route, tiles='openstreetmap')

orig_loc = (start[0],start[1])
dest_loc = (end[0],end[1])
orig_marker = folium.Marker(
            location = orig_loc,
            popup = 'Boston Public Garden',
            icon = folium.Icon(color='red'))
dest_marker = folium.Marker(
            location = dest_loc,
            popup = 'Massachusetts General Hospital',
            icon = folium.Icon(color='green'))
orig_marker.add_to(route_map)
dest_marker.add_to(route_map)

route_map
```
